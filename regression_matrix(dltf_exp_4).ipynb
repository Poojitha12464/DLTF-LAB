```python
import numpy as np;
y_true=np.array([50, 62, 45, 70, 55])
y_pred=np.array([52, 50, 49, 69, 65])
errors=y_true-y_pred
mae=np.mean(np.abs(errors))
print(f"MAE: {mae:.2f}")
squared_errors=np.square(errors)
mean_squared_error=np.mean(squared_errors)
print(f"MSE: {mean_squared_error:.2f}")
rmse=np.sum(squared_errors)/len(squared_errors)
print(f"RSME: {rmse:.2f}")
ss_res=np.sum(squared_errors)
y_mean=np.mean(y_true)
ss_tot=np.sum(np.square(y_true-y_mean))
r2=1-(ss_res/ss_tot)
print(f"R^2: {r2:.2f}")
```
